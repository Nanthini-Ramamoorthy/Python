```python
# Body Mass Index
weight = float(input("Enter the BMI Index : "))
if(weight < 18.5):
    print("Under weight")
elif(weight <= 24.9):
    print("Normal weight")
elif(weight <= 29.9):
    print("Over weight")
elif(weight <= 34.9):
    print("Very Over weight")
elif(weight >= 35.0):
    print("Extremely Obese")
```
